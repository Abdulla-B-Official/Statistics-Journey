## Question - 1  
The historical average score in a certification exam is 65.A training institute introduces a new course. A random sample of 50 students trained under the new course has an average score of 68.At the 5% level of significance, test whether the new course improves performance.

```python
from scipy import stats
import numpy as np

# Generate sample data
observations_40_tickets = np.random.normal(loc=33, scale=5.0, size=40).astype(int)

# One-sample right-tailed t-test
t_stats, p_value = stats.ttest_1samp(a=observations_40_tickets, popmean=30, alternative='greater')

print("t-statistic:", t_stats)
print("p-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject H0")
else:
    print("Do not Reject H0")
```

### Final conclusion :

| Item      | Answer                                   |
| --------- | ---------------------------------------- |
| Test type | One-sample t-test                        |
| Tail type | One-tailed (right-tailed)                |
| Reason    | Testing if mean **increased** ( μ > 30 ) |


## Question - 2
A company claims that the average time taken to resolve customer tickets is 30 minutes.A random sample of 40 tickets shows a mean resolution time of 33 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
import numpy as np
from scipy import stats

# Sample data (example: resolution times around mean 33)
sample_data = np.random.normal(loc=33, scale=5, size=40)

# Population mean from company claim
mu = 30

# One-sample right-tailed t-test
t_stat, p_value = stats.ttest_1samp(sample_data, popmean=mu, alternative='greater')

print("t-statistic:", t_stat)
print("p-value:", p_value)

# Decision rule at 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Reject H0: Average resolution time has increased.")
else:
    print("Fail to reject H0: No evidence that resolution time increased.")
```

### Final conclusion :

| Item      | Answer                                   |
| --------- | ---------------------------------------- |
| Test type | One-sample t-test                        |
| Tail type | One-tailed (right-tailed)                |
| Reason    | Testing if mean **increased** ( μ > 30 ) |

## Question - 3
A company claims that 60% of its users are satisfied with its product.
In a survey of 200 users, 134 users report satisfaction.
At the 5% level of significance, test whether customer satisfaction is higher than claimed.


```python
!pip install statsmodels
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Given values
satisfied_users = 134
total_users = 200
claimed_proportion = 0.60

# Perform one-sample proportion z-test
z_stat, p_value = proportions_ztest(count=satisfied_users,
                                    nobs=total_users,
                                    value=claimed_proportion,
                                    alternative='larger')

print("Z-statistic:", z_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: Customer satisfaction is higher than claimed.")
else:
    print("Fail to reject H0: No evidence that satisfaction is higher.")
```

### Final conclusion :
| Item      | Answer                           |
| --------- | -------------------------------- |
| Test      | One-sample proportion test       |
| Statistic | Z-test                           |
| Tail      | One-tailed (right-tailed)        |
| Reason    | Testing if proportion **> 0.60** |

## Question - 5
Two teaching methods are used to train employees.
A sample of 30 employees trained using Method A and 28 employees trained using Method B are tested.
Test at the 5% level of significance whether there is a difference in average scores between the two methods.


```python
import numpy as np
from scipy import stats

# Example sample scores
method_A = np.random.normal(loc=75, scale=8, size=30)
method_B = np.random.normal(loc=72, scale=8, size=28)

# Two-sample t-test (two-tailed)
t_stat, p_value = stats.ttest_ind(method_A, method_B)

print("t-statistic:", t_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: There is a significant difference between the two methods.")
else:
    print("Fail to reject H0: No significant difference between the methods.")
```

### Final conclusion
| Item      | Answer                                                                |
| --------- | --------------------------------------------------------------------- |
| Test      | Two-sample t-test                                                     |
| Statistic | t-test                                                                |
| Tail      | Two-tailed                                                            |
| Reason    | Testing whether the **means of two independent groups are different** |

## Question - 7
Two versions of a website are tested.
Out of 500 users for Version A, 320 users click the signup button.
Out of 480 users for Version B, 290 users click the signup button.
At the 5% level of significance, test whether the conversion rates differ.


```python
from statsmodels.stats.proportion import proportions_ztest

# Data
clicks = [320, 290]        # number of successes
users = [500, 480]         # total users

# Two-proportion z-test
z_stat, p_value = proportions_ztest(clicks, users)

print("Z-statistic:", z_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: Conversion rates differ.")
else:
    print("Fail to reject H0: No significant difference in conversion rates.")
```

### Final conclusion :
| Item      | Answer                                                                 |
| --------- | ---------------------------------------------------------------------- |
| Test      | Two-proportion Z-test                                                  |
| Statistic | Z-test                                                                 |
| Tail      | Two-tailed                                                             |
| Reason    | Testing if **conversion rates differ between Version A and Version B** |

## Question - 9
A company wants to compare the average productivity of employees trained using three different training programs.
At the 5% level of significance, test whether there is a difference in mean productivity across the programs.


```python
import numpy as np
from scipy import stats

# Example productivity scores
program_A = np.random.normal(loc=75, scale=6, size=25)
program_B = np.random.normal(loc=78, scale=6, size=25)
program_C = np.random.normal(loc=82, scale=6, size=25)

# One-way ANOVA test
f_stat, p_value = stats.f_oneway(program_A, program_B, program_C)

print("F-statistic:", f_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: Mean productivity differs across programs.")
else:
    print("Fail to reject H0: No significant difference in productivity.")
```

### Final conclusion :
| Item      | Answer                                         |
| --------- | ---------------------------------------------- |
| Test      | One-way ANOVA                                  |
| Statistic | F-test                                         |
| Reason    | Comparing **means of three training programs** |


## Question - 10
A company wants to test whether gender and product preference are independent.
A contingency table is created using survey data.
At the 5% level of significance, test for independence.


```python
import numpy as np
from scipy.stats import chi2_contingency

# Example contingency table
# Rows: Gender (Male, Female)
# Columns: Product preference (Product A, Product B, Product C)

data = np.array([[50, 30, 20],
                 [45, 35, 25]])

# Chi-square test
chi2_stat, p_value, dof, expected = chi2_contingency(data)

print("Chi-square statistic:", chi2_stat)
print("p-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: Gender and product preference are not independent.")
else:
    print("Fail to reject H0: No evidence of dependence.")
```

### Final conclusion :
| Item      | Answer                                                            |
| --------- | ----------------------------------------------------------------- |
| Test      | Chi-Square Test of Independence                                   |
| Statistic | χ²                                                                |
| Reason    | Testing whether **gender and product preference are independent** |

## Question - 12
A dice manufacturer claims that a die is fair.
The die is rolled 120 times, and observed frequencies are recorded.
Test at the 5% level of significance whether the die is fair.


```python
import numpy as np
from scipy.stats import chisquare

# Example observed frequencies for die faces (1–6)
observed = np.array([18, 22, 17, 21, 20, 22])

# Expected frequencies for a fair die
expected = np.array([20, 20, 20, 20, 20, 20])

# Chi-square goodness-of-fit test
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print("Chi-square statistic:", chi2_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject H0: The die is not fair.")
else:
    print("Fail to reject H0: No evidence that the die is unfair.")
```

### Final conclusion :
| Item      | Answer                                                              |
| --------- | ------------------------------------------------------------------- |
| Test      | Chi-Square Goodness-of-Fit                                          |
| Statistic | χ²                                                                  |
| Reason    | Testing whether **observed die outcomes match a fair distribution** |
